## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

I'm going with Lane 4 CTR/Engagement Opportunity Scoring. Some pages rank well in search but still don't get clicked as much as similar-ranking pages do. That's usually a title or meta-description problem not a content problem and it's cheap to fix compared to a full rewrite so it's worth flagging separately.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Action: the content team rewrites the snippet for flagged pages
Cost of a wrong call: if you flag the wrong pages, you waste rewrite effort on pages that were never underperforming or you miss genuinely bad pages and leave clicks on the table

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

# Only look at pages ranking well
good_position = df[df["position_tier"].isin(["top_3", "page_1"])]
print("Pages ranking top_3 or page_1:", len(good_position))

# Apply a volume floor - low-impression pages give noisy CTR numbers
reliable = good_position[good_position["impressions_90d"] >= 100].copy()
print("Of those, with impressions_90d >= 100:", len(reliable))

# Compare each page's CTR to its tier's average CTR
tier_avg_ctr = reliable.groupby("position_tier")["ctr"].mean()
reliable["tier_avg_ctr"] = reliable["position_tier"].map(tier_avg_ctr)
underperformers = reliable[reliable["ctr"] < reliable["tier_avg_ctr"]]

print("Underperforming candidates:", len(underperformers))
print(tier_avg_ctr)
```

Out of 30,000 total pages, 14,135 rank well enough to matter for this lane (top_3 or page_1). After applying a volume floor of at least 100 impressions in the last 90 days to avoid noisy CTR numbers from low-traffic pages 9,166 pages remain reliable to compare. Of those 6,060 pages have a CTR below their tier's average, meaning about two-thirds of well-ranking pages are underperforming relative to peers at the same position. Average CTR was actually slightly higher for page_1 (35.5%) than top_3 (33.4%) in this slice which is worth flagging rather than assuming top_3 always wins.
