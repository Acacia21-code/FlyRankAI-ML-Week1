## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a scoring task not classification or clustering the goal is to produce a continuous opportunity number for each page not sort pages into fixed categories.
Note: the data already has is_declining_label a pre-built classification target for a different lane (decline detection). I am deliberately not using it here.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

There is no direct opportunity label in the data so I am using a proxy: how far a page's actual CTR falls below the typical CTR for pages at the same position_tier. A page ranking well (e.g. top_3) but getting fewer clicks than similar pages is the opportunity signal.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@50 of the top 50 pages the model flags as the highest opportunity how many turn out to actually have a real CTR gap. Calculated only on pages passing 'measurable_opportunity' (impressions_90d > 100 and sessions_90d > 0), so low-traffic noise does not distort the metric  

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
df.head()
```

One row = one page, identified by 'content_id' (a pseudonymous ID, unique per row used only for joins never as a model feature).

```python
import numpy as np

df['expected_ctr'] = df.groupby('position_tier')['ctr'].transform('median')
df['opportunity_score'] = (df['expected_ctr'] - df['ctr']).clip(lower=0) * np.log1p(df['impressions_90d'])

df[['content_id', 'position_tier', 'ctr', 'expected_ctr', 'opportunity_score']].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A flat rule like "flag anything under X% CTR" ignores that normal CTR varies a lot by position_tier, main_intent and content type a fixed threshold would misfire across differe query types. A model can weigh avg_position, search_volume, competition_level, engagement-rate and trend_direction together and catch interactions a single rule would miss like ai_traffic_pct pulling sessions away from organic search which makes raw CTR misleading on its own.
